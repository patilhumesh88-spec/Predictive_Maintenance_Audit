```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_PATH = Path("../data/raw/ai4i2020.csv")

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
df.head()
```

```python
missing = df.isna().sum()
print(missing)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
for col in df.columns:
    print(f"\n--- {col} ---")
    print("Unique values:", df[col].nunique())
    print(df[col].unique()[:10])
```

```python
target_counts = df["Machine failure"].value_counts()

print(target_counts)

print("\nPercentages:")
print(df["Machine failure"].value_counts(normalize=True) * 100)
```

```python
mode_cols = ["TWF", "HDF", "PWF", "OSF", "RNF"]

for col in mode_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

```python
mode_sum = df[mode_cols].sum(axis=1)

print("Rows with at least one failure mode:",
      (mode_sum > 0).sum())

print("Machine failure rows:",
      df["Machine failure"].sum())

print("Mode present but Machine failure = 0:",
      ((mode_sum > 0) & (df["Machine failure"] == 0)).sum())

print("Machine failure = 1 but no mode flagged:",
      ((mode_sum == 0) & (df["Machine failure"] == 1)).sum())
```

```python
print("UDI unique:", df["UDI"].nunique(), "/", len(df))

print("\nProduct ID unique:", df["Product ID"].nunique())
print(df["Product ID"].value_counts().head(10))
```
